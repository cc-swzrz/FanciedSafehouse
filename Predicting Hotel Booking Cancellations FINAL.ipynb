# Predicting Hotel Booking Cancellations

## Defining the Challenge

Hotels experience a significant number of booking cancellations, leading to revenue losses and inefficiencies in resource planning.

**Goals:**
- Develop a machine learning model that can predict whether a booking will be cancelled based on the data provided.
- Identify key factors contributing to cancellations.

### Data Preprocessing & Feature Engineering

```python
#Check for null values
hotel_reservations_df.isnull().sum()
```

```python
#check number of unique values in each column
hotel_reservations_df.nunique()
```

```python
#drop Booking_ID
hotel_reservations_df.drop(['Booking_ID'], axis='columns', inplace=True)
```

```python
#drop arrival_year
hotel_reservations_df.drop(['arrival_year'], axis='columns', inplace=True)
```

```python
#One-Hot encoding for type_of_meal_plan, room_type_reserved, market_segment_type
hotel_reservations_df = pd.get_dummies(hotel_reservations_df, 
                                       columns=['type_of_meal_plan', 'room_type_reserved', 'market_segment_type'], 
                                       drop_first=True)
```

```python
#label encode booking_status

#create a LabelEncoder object
label_encoder = LabelEncoder()

hotel_reservations_df['booking_status'] = label_encoder.fit_transform(hotel_reservations_df['booking_status'])
hotel_reservations_df.head()
```

```python
#check correlation between features in df again
num_cols = hotel_reservations_df.select_dtypes(exclude='object')
corr = num_cols.corr()

#creating a heat map of all the numerical features
fig = px.imshow(corr)
fig.update_xaxes(side="top")

#adjust the size of the heatmap
fig.update_layout(
    width=1000,  # Adjust the width
    height=800   # Adjust the height
)
fig.show()
```

```python
#drop type_of_meal_plan_not_selected', 'type_of_meal_plan_Meal Plan 3', 'room_type_reserved_Room_Type 3
hotel_reservations_df = hotel_reservations_df.drop(columns=[
    'type_of_meal_plan_Not Selected', 
    'type_of_meal_plan_Meal Plan 3', 
    'room_type_reserved_Room_Type 3'
])
```

## Model Selection and Training

```python
#Split the data into training and test sets
X = hotel_reservations_df.drop(columns=['booking_status'])
y = hotel_reservations_df['booking_status']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=5)
```

```python
#Instantiate the model object
xgb_model = XGBClassifier(learning_rate=0.1, random_state=24, max_depth=18)
rf_model = RandomForestClassifier(n_estimators=100, random_state=24)
```

```python
#Fit the model with the training data
xgb_model.fit(X_train, y_train)
rf_model.fit(X_train, y_train)
```

```python
#predict the target on the test dataset
xgb_y_predict = xgb_model.predict(X_test)
rf_y_predict = rf_model.predict(X_test)
```

```python
#Calculate the Accuracy and F1 score
print("Test Scores for XGBoost")
accuracy_test = metrics.accuracy_score(y_test, xgb_y_predict)
print('Accuracy_score: ', accuracy_test)
f1 = f1_score(y_test, xgb_y_predict)
print("F1 Score: ", f1)

print("\nTest Scores for Random Forest")
accuracy_test = metrics.accuracy_score(y_test, rf_y_predict)
print('Accuracy_score: ', accuracy_test)
f1 = f1_score(y_test, rf_y_predict)
print("F1 Score: ", f1)
```

```python
#get feature importances
feature_importances = rf_model.feature_importances_

#creating a dataframe
feature_df = pd.DataFrame({
    'Feature': X_train.columns,
    'Importance': feature_importances
})
# Sort by importance
feature_df = feature_df.sort_values(by='Importance', ascending=False)

# Plot the feature importances
plt.figure(figsize=(10, 6))
plt.barh(feature_df['Feature'], feature_df['Importance'])
plt.xlabel('Feature Importance')
plt.ylabel('Feature')
plt.title('Feature Importance from Random Forest')
plt.show()
```
